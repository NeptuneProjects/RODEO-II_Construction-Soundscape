```python
from pathlib import Path

import matplotlib.pyplot as plt
import tomllib

import polars as pl

from rodeo import plotting

plt.style.use(Path("config/jasa_fig_style.mplstyle"))
```

```python
bounds = [[-71.25, -71.0], [41.05, 41.15]]
# Pile and sensor positions from the South Fork Wind site configuration.
with open("config/sites/southfork.toml", "rb") as f:
    site = tomllib.load(f)
piles = pl.read_csv(site["piles"])
sensors = pl.DataFrame(
    [{k: s[k] for k in ("name", "latitude", "longitude")} for s in site["sensors"]]
)
```

```python
fig = plotting.plot_study_area(bounds, piles, sensors)
fig.savefig("results/study_area.pdf", dpi=600, bbox_inches="tight", pad_inches=0.0)
plt.show()
```
